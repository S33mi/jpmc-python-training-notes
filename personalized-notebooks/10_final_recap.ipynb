```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

print("🎯 Final Recap Showcase - Running Live Analysis...\n")

# Fetch sample Stocks + Global data
tickers = ['JPM', 'TSLA', 'AAPL', 'MSFT']
data = yf.download(tickers, start="2025-01-01", progress=False)['Close']
returns = data.pct_change().dropna()

# 1. Summary Table
summary = pd.DataFrame({
    'Annual Return': returns.mean() * 252,
    'Annual Vol': returns.std() * np.sqrt(252),
    'Sharpe Ratio': (returns.mean() * 252) / (returns.std() * np.sqrt(252))
}).round(4)

print("📊 Performance Summary:")
display(summary)

# 2. Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(returns.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation of Daily Returns')
plt.show()

# 3. Cumulative Returns
(returns + 1).cumprod().plot(figsize=(10, 5), linewidth=2)
plt.title('Cumulative Returns – Global Stocks')
plt.ylabel('Growth of $1')
plt.grid(True, alpha=0.3)
plt.show()
```
