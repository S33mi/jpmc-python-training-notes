## Lists – Handling Multiple Values (e.g., Stock Prices Over Time)

```python
# Create a list of daily closing prices
prices = [250.5, 252.0, 248.75, 255.10, 253.40]
print(prices)
print(type(prices))          # <class 'list'>
print(len(prices))           # 5 days
```

```python
# Access elements (index starts at 0)
print(prices[0])             # First price: 250.5
print(prices[-1])            # Last price: 253.40 (act as a circular linked list)
```

```python
# Add / modify
prices.append(260.0)         # New day
prices[2] = 249.0            # Correct error at 3rd price
print(prices)
```

```python
# Calculate simple returns between days
returns = []
for i in range(1, len(prices)):
    ret = (prices[i] - prices[i-1]) / prices[i-1] * 100
    returns.append(round(ret, 2))
print("Daily returns (%):", returns)
```

## Loops – Repeating Actions (e.g., Process Many Transactions)

```python
# For loop over list
for price in prices:
    print(f"Price: {price:.2f} USD")

# With index (useful for dates)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
for i, day in enumerate(days):
    print(f"{day}: {prices[i]:.2f}")
```

```python
# While loop example (rare but good to know)
balance = 10000
interest_rate = 0.07 / 12   # Monthly
months = 0
while balance < 15000:   #counter condition
    balance *= (1 + interest_rate)    # you need to adjust (increse or decrese) counter condition on your own
    months += 1
print(f"Time to reach 15,000 USD: {months} months")
```

## Functions – Reusable Code Blocks (e.g., Calculate Returns)

```python
def calculate_return(initial, final):
    """Simple percentage return"""
    return (final - initial) / initial * 100

print(calculate_return(10000, 11500))          # 15.0
print(calculate_return(500, 450))              # -10.0
```

```python
# More useful: With optional parameters
def portfolio_value(shares, price, currency="USD"):
    total = shares * price
    return f"{total:,.2f} {currency}"

print(portfolio_value(120, 250.75))            # 30,090.00 USD
print(portfolio_value(50, 180.0, "USD"))       # 9,000.00 USD
```
